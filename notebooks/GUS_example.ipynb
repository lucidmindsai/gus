```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
# import seaborn as sns

import math
import random
import logging
from typing import Union

# from shapely.geometry import Polygon, MultiPolygon, Point

from pygus import Urban, Species, tree_population_from_geojson, ScenarioConfig, SiteConfig, WeatherConfig

print(os.getcwd())
ALLOMETRICS = Species(os.path.join("..", "pygus", "inputs", "allometrics.json"))
```

```python
population = pd.read_csv("test_pop.csv")
population.head()
```

```python
Weather = WeatherConfig(growth_season_mean=153, growth_season_var=7)

site_file ="city_config.json"
site = SiteConfig(total_m2=1000,
                  impervious_m2=0,
                  pervious_m2=1000,
                  weather=Weather,
                  light_loss_multiplier=0.1)
```

```python
scenario_low = ScenarioConfig(time_horizon_years=50, maintenance_scope = 1, contagion_level = 'low')
scenario_high = ScenarioConfig(time_horizon_years=50, maintenance_scope = 1, contagion_level = 'high')
```

```python
proj_low = Urban(population, species_allometrics=ALLOMETRICS, site_config=site, scenario=scenario_low)
```

```python
proj_high = Urban(population, species_allometrics=ALLOMETRICS, site_config=site, scenario=scenario_high)
```

```python
outputs_low = proj_low.run_model()
```

```python
outputs_high = proj_high.run_model()
```
